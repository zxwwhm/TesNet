```python
import os
import json

combine_file = './30shot_combine.json'
file_dir = '../seed0'
data_dict = {}
with open('../seed0/full_box_30shot_person_trainval.json', 'r') as load_f:
    file = json.load(load_f)
    data_dict['info'] =file['info']
    data_dict['licenses'] = file['licenses']
    data_dict['categories']=file['categories']
    print(file.keys())

data_dict['images'] = []
data_dict['annotations'] =[]
i=0
for path,dir_list,file_list in os.walk(file_dir):
    for file in file_list:
        for name in ('person','airplane','bicycle','bird','boat','bottle','bus','car','cat','chair','couch','cow','dining table','dog'
                               ,'horse','motorcycle','potted plant','sheep','train','tv'):
            name = name + '_'
            if '30shot' in file and name in file:
                print(file, name)
                with open(os.path.join(file_dir,file),'r') as load_f:
                    file =json.load(load_f)
                    for image in file['images']:
                        data_dict['images'].append(image)
                        i=i+1
                    print(i)
                    for ann in file['annotations']:
                        data_dict['annotations'].append(ann)

with open(combine_file,"w") as dump_f:
    json.dump(data_dict,dump_f)
```
